```python
import pygame

import squirrel_config as config
import random
from squirrel_function import (
    Button, draw_player, load_player_image, Enemy, game_over_screen, 
    you_win_screen
)

def main():
    pygame.init()
    screen = pygame.display.set_mode([config.WIDTH, config.HEIGHT])
    pygame.display.set_caption("Apple Orchard Collect")

    clock = pygame.time.Clock()
    font = pygame.font.Font("freesansbold.ttf", 34)

    
    button_width = 300
    button_height = 90
    center_x = (config.WIDTH - button_width) // 2
    center_y = (config.HEIGHT - button_height) // 2

    play_button = Button("Play!", center_x, center_y, button_width, button_height, True)

    display_text = font.render("Welcome to Apple Orchard Collect!", True, config.BOX_COLOR)
    text_rect = display_text.get_rect(center=(config.WIDTH // 2, center_y - 50))
    screen.blit(display_text, text_rect)    
    play_button.draw(screen, font)

    state = "menu"

    player_x = 50
    player_y = 50
    player_rect = pygame.Rect(player_x, player_y, config.PLAYER_IMAGE_SIZE[0], config.PLAYER_IMAGE_SIZE[1])
    blocked = False

    apple_image = pygame.image.load(config.APPLE_IMAGE_PATH).convert_alpha()
    apple_image = pygame.transform.scale(apple_image, config.APPLE_SIZE)

    enemies = [Enemy(random) for _ in range(6)]

    score = 0

    running = True
    
    while running:
        dt = clock.tick(config.FPS) / 1000.0

        keys = pygame.key.get_pressed()
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False

            if event.type == pygame.MOUSEBUTTONDOWN:
                if state == "menu" and play_button.is_clicked(pygame.mouse.get_pos()):
                    state = "game"

        if state == "menu":
            screen.fill(config.PLAY_SCREEN)
            play_button.draw(screen, font)
            display_text = font.render("Welcome to Apple Orchard Collect!", True, config.TEXT_COLOR)
            text_rect = display_text.get_rect(center=(config.WIDTH // 2, center_y - 50))
            screen.blit(display_text, text_rect)

        elif state == "game":
            screen.fill(config.BG_COLOR)

            load_player_image()
            
            keys = pygame.key.get_pressed()
            dx = dy = 0
            if keys[pygame.K_LEFT] or keys[pygame.K_a]:dx = -1   
            if keys[pygame.K_RIGHT] or keys[pygame.K_d]: dx = 1
            if keys[pygame.K_UP] or keys[pygame.K_w]: dy = -1
            if keys[pygame.K_DOWN] or keys[pygame.K_s]: dy = 1
            
            if dx != 0 and dy != 0:
                dx *= 0.7071
                dy *= 0.7071

            new_x = player_x + dx * config.PLAYER_SPEED * dt
            new_y = player_y + dy * config.PLAYER_SPEED * dt
            
            
            if not blocked: #not is_blocked(new_x, new_y):
                player_x, player_y = new_x, new_y
        
            
            player_rect.center = (player_x, player_y)

            for apple in config.APPLES.values():
                apple_rect = pygame.Rect(apple["pos"], config.APPLE_SIZE)
                screen.blit(apple_image, apple["pos"])
                
                if player_rect.colliderect(apple_rect):
                    score += 1
                    apple["pos"] = (-200, -200)  
                if score == 10:
                    state = "win"
                    
            
            for enemy in enemies:
                enemy.update()
                enemy.draw(screen)

                if player_rect.colliderect(pygame.Rect(enemy.x, enemy.y, enemy.size, enemy.size)):
                    state = "lose"

            draw_player(screen, player_x, player_y)
            score_text = font.render(f"Score: {score}", True, config.TEXT_COLOR) 
            screen.blit(score_text, (650,20))

        elif state == "lose":
            screen.fill((0,0,0))
            game_over_screen(screen)
            

        elif state == "win":
            screen.fill((0,0,0))
            you_win_screen(screen)
    
        pygame.display.flip()
    
    pygame.time.wait(200)
    clock.tick(60)

    pygame.quit()
if __name__ == "__main__":
    main()
```
